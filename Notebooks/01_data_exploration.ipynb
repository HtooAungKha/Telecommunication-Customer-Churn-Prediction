#### Outlier Analysis

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

plt.figure(figsize=(12, 5))

for index, column in enumerate(numerical_columns, start=1):
    plt.subplot(1, 3, index)
    sns.boxplot(y=df[column])
    plt.title(column)

plt.tight_layout()
plt.show()
```

#### Count Outliers Using the IQR Method

```python
for column in numerical_columns:
    clean_values = df[column].dropna()

    Q1 = clean_values.quantile(0.25)
    Q3 = clean_values.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = clean_values[
        (clean_values < lower_bound) |
        (clean_values > upper_bound)
    ]

    print(f"{column}: {len(outliers)} outliers")
```

#### Outlier Analysis

The boxplots and IQR method were used to check the numerical features. No outliers were detected in `tenure`, `MonthlyCharges`, or `TotalCharges` using the IQR rule. The numerical values fall within the expected range, so no records were removed before preprocessing.
